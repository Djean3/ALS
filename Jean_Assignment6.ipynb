# Setup

```python

from pyspark.sql import SparkSession

spark = SparkSession.builder \
    .appName("Assignment6") \
    .master("local[*]") \
    .config("spark.driver.host", "127.0.0.1") \
    .getOrCreate()

# 📦 Initialize the RDD
rdd_data = spark.sparkContext.parallelize([
    "this is an example of spark",
    "the example contains several lines",
    "spark has low level and high level api"
])
```

# Question 1

```python
# Q1
rdd_split = rdd_data.flatMap(lambda element: element.split(" "))
print("rdd_split:", rdd_split.collect())

```

- Explanation: splits each line into words, then flattens into a single RDD of words.

# Question 2

```python
rdd_map = rdd_split.map(lambda element: (element, 1))
print("rdd_map:", rdd_map.collect())
```

- Explanation: Each word is paired with 1 — used for counting word frequency.

# Question 3

```python
rdd_reduce = rdd_map.reduceByKey(lambda x, y: x + y)
print("rdd_reduce:", rdd_reduce.collect())
```

- Explanation: All counts for the same word are added together

# Question 4

```python
rdd_filter = rdd_reduce.filter(lambda element: 'r' in element[0])
print("rdd_filter:", rdd_filter.collect())
```

- Explanation: Filters for words containing the letter 'r' only

# Question 5

```python
# Q5
myData = "resilient distributed datasets".split(" ")
data = spark.sparkContext.parallelize(myData, 2)
print("data.collect():", data.collect())
```

- Explanation: sparkContext.parallelize() turns a Python list into an RDD with 2 partitions

# Question 6

```python
distinctOut = data.distinct()
print("distinctOut.count():", distinctOut.count())
```

- Explanation: Returns number of unique elements in the RDD

# Question 7


```python
dataMap = data.map(lambda element: (element, element[0], element.startswith("d")))
filtered = dataMap.filter(lambda record: record[2]).take(5)
print("Filtered records:", filtered)
```

- Explanation: Tuple contains (word, first letter, starts_with_d) and filters for `True` on third item
